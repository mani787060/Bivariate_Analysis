# 1.scatterplot(Numerical-Numerical)

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip']) #bivariate analysis
```

```python
# multivariate_analysis
sns.scatterplot(x=tips['total_bill'], y=tips['tip'],hue=tips['sex'], style=tips['smoker'],size=tips['size']) 
```

# 2.Barplot(Numerical-Categorical)

```python
# keep numerical column always on y-axis
# keep categorical column always on x-axis
sns.barplot(x=titanic['Pclass'], y=titanic['Age']) # bivariate
```

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'], hue=titanic['Sex'])
```

# 3.Boxplot(Numerical-Categorical)

```python
sns.boxplot(x=titanic['Pclass'],y=titanic['Age'], hue=titanic['Sex'])
```

# 4. Distplot(Numerical-Categorical)

```python
# person who dead in titanic
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False) 
# person who survived in titanic
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist=False)
```

# 5.Heatmap(Categorical-Categorical)

```python
data = pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
sns.heatmap(data)
```

```python
titanic.groupby('Pclass')['Survived'].mean()*100
```

```python
# you can check this data by using any graph(bar, pie etc...)
(titanic.groupby('Pclass')['Survived'].mean()*100).plot(kind='pie', autopct='%.2f')
```

# 6. Clustermap(Categorical-Categorical)

```python
pd.crosstab(titanic['Parch'], titanic['Survived'])
```

```python
# shows similarity between the data
sns.clustermap(pd.crosstab(titanic['Parch'], titanic['Survived']))
```

# 7.Pairplot(collection of scatterplot)

```python
sns.pairplot(iris, hue='species')
```

# 8. Lineplot(Numerical-Numerical)

```python
#flights.groupby(['Year','Month'])['Passengers']
new = flights.groupby(['Year','Month']).sum().reset_index()
```

```python
sns.lineplot(x=new['Year'],y=new['Passengers'], data=new)
```

```python
flights.pivot_table(values='Passengers', index='Month', columns='Year').astype(int)
```

```python
sns.heatmap(flights.pivot_table(values='Passengers', index='Month', columns='Year'))
```

```python
sns.clustermap(flights.pivot_table(values='Passengers', index='Month', columns='Year'))
```
